<h1>Problem 16</h1>
What is the sum of the digits of the number $2^{1000}$?

```python
num = 2**1000
digits = str(num)
digit_sum = 0
for i in range(0,len(digits)):
    digit_sum += int(digits[i])
print(digit_sum)
```

<h1>Problem 17</h1>
If all the numbers from 1 to 1000 were written out in words, how many letters would be used? (Exclude spaces and hyphens)

```python
def calc_chars(num):
    
    num_letters = [3,3,5,4,4,3,5,5,4,3,6,6,8,8,7,7,9,8,8,6,6,5,5,5,7,6,6,7,8]
    # represents [one,two,three,...,twenty,thirty,forty,...,hundred,thousand]
    
    letter_sum = 0
    for i in range(1, num + 1):
        number = str(i)
        
        if number[-1] != str(0):
            letter_sum += num_letters[int(number[-1]) - 1]

        if len(number) > 1:
            if number[-2] == str(1):
                letter_sum += num_letters[int(number[-2:]) - 1]
                if number[-2:] != str(10):
                    letter_sum -= num_letters[int(number[-1]) - 1] # cancels out first line
            elif number[-2] != str(0):
                letter_sum += num_letters[int(number[-2]) + 17] # this lines up since the 19th (17+2) index = 20
            
            if len(number) > 2:
                if int(number) % 100 != 0: # condition for 'and'
                        letter_sum += 3
                if number[-3] != str(0):
                    letter_sum += (num_letters[int(number[-3]) - 1] + num_letters[-2]) # [-2] must be changed if larger numbers are added to num_letters
                if len(number) > 3 and number[-4] != str(0):
                    letter_sum += (num_letters[int(number[-4]) - 1] + num_letters[-1])
                            
            
    return letter_sum

calc_chars(1000)
```

```python
jan_1_1900 = 1 # Monday
jan_1_1901 = (jan_1_1900 + 365) % 7
# 1st of the months for non leap years are [0,31,59,90,120,151,181,212,243,273,304,334]
# 1st of the months for leap years are [0,31,60,91,121,152,182,213,244,274,305,335]
prev_year = []
for m in range(365):
    prev_year.append((jan_1_1901 + m) % 7)
curr_year = []
sundays = 0
for i in range(1901,2001):
    if i % 4 == 0 and (i % 100 != 0 or i % 400 == 0): # change this
        for j in range(366):
            curr_year.append(((prev_year[-1] + 1 + j) % 7))
        if curr_year[0] == 0:
            sundays += 1
        if curr_year[31] == 0:
            sundays += 1
        if curr_year[60] == 0:
            sundays += 1
        if curr_year[91] == 0:
            sundays += 1
        if curr_year[121] == 0:
            sundays += 1
        if curr_year[152] == 0:
            sundays += 1
        if curr_year[182] == 0:
            sundays += 1
        if curr_year[213] == 0:
            sundays += 1
        if curr_year[244] == 0:
            sundays += 1
        if curr_year[274] == 0:
            sundays += 1
        if curr_year[305] == 0:
            sundays += 1
        if curr_year[335] == 0:
            sundays += 1
                
    else: # not leap year
        for j in range(365):
            curr_year.append(((prev_year[-1] + 1 + j) % 7))
        if curr_year[0] == 0:
            sundays += 1
        if curr_year[31] == 0:
            sundays += 1
        if curr_year[59] == 0:
            sundays += 1
        if curr_year[90] == 0:
            sundays += 1
        if curr_year[120] == 0:
            sundays += 1
        if curr_year[151] == 0:
            sundays += 1
        if curr_year[181] == 0:
            sundays += 1
        if curr_year[212] == 0:
            sundays += 1
        if curr_year[243] == 0:
            sundays += 1
        if curr_year[273] == 0:
            sundays += 1
        if curr_year[304] == 0:
            sundays += 1
        if curr_year[334] == 0:
            sundays += 1
            
    prev_year = curr_year
    curr_year = []

print(sundays)
```

<h1>Problem 20</h1>
Find the sum of the digits in the number 100!

```python
def factorial(n):
    if n == 0: #base case
        return 1
    else:
        return n * factorial(n-1)
    
sum = 0
digits = str(factorial(100))
for x in digits:
    sum += int(x)
print(sum)
```
