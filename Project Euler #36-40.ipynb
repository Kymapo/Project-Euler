## Problem 36

The decimal number, 585 = 1001001001(binary), is palindromic in both bases.

Find the sum of all numbers, less than one million, which are palindromic in base 10 and base 2.

(Please note that the palindromic number, in either base, may not include leading zeros.)

```python
def test_palindromes():

  tot_sum = 0
  for a in range(10):
    for b in range(10):
      for c in range(10):
          num = 10000*a + 1000*b + 100*c + 10*b + a
          while (num % 10) == 0 and num != 0:
            num = num // 10

          powers_of_2 = []
          k=0
          while 2**k <= num:
            powers_of_2.append(2**k)
            k+=1
          bin_seq = []
          rem = num


          for m in range(len(powers_of_2)):
            if rem >= powers_of_2[-(m+1)]:
              bin_seq.append(1)
              rem = rem - powers_of_2[-(m+1)]
            else:
              bin_seq.append(0)
          if bin_seq == bin_seq[::-1]:
            tot_sum += num
            #print(num, bin_seq)

  for a in range(10):
    for b in range(10):
      for c in range(10):
          num = 100000*a + 10000*b + 1000*c + 100*c + 10*b + a
          while (num % 10) == 0 and num != 0:
            num = num // 10

          powers_of_2 = []
          k=0
          while 2**k <= num:
            powers_of_2.append(2**k)
            k+=1
          bin_seq = []
          rem = num


          for m in range(len(powers_of_2)):
            if rem >= powers_of_2[-(m+1)]:
              bin_seq.append(1)
              rem = rem - powers_of_2[-(m+1)]
            else:
              bin_seq.append(0)
          if bin_seq == bin_seq[::-1]:
            tot_sum += num
            #print(num, bin_seq)

  return tot_sum

test_palindromes()
```

##Problem 38

<p>Take the number $192$ and multiply it by each of $1$, $2$, and $3$:</p>
\begin{align}
192 \times 1 &= 192\\
192 \times 2 &= 384\\
192 \times 3 &= 576
\end{align}
<p>By concatenating each product we get the $1$ to $9$ pandigital, $192384576$. We will call $192384576$ the concatenated product of $192$ and $(1,2,3)$.</p>
<p>The same can be achieved by starting with $9$ and multiplying by $1$, $2$, $3$, $4$, and $5$, giving the pandigital, $918273645$, which is the concatenated product of $9$ and $(1,2,3,4,5)$.</p>
<p>What is the largest $1$ to $9$ pandigital $9$-digit number that can be formed as the concatenated product of an integer with $(1,2, \dots, n)$ where $n \gt 1$?</p>


```python
def pandigital_product():
  pandigitals = [918273645]
  for i in range(20,25):
    if i % 5 == 0:
      continue
    twice = i * 2
    thrice = i * 3
    fource = i * 4
    fivce = i * 5
    digits = str(i) + str(twice) + str(thrice) + str(fource) + str(fivce)

    if len(digits) != 9:
      continue
    if len(set(digits)) == len(digits):
      pandigitals.append(int(digits))

  for i in range(100,333):
    if i % 5 == 0:
      continue
    twice = i * 2
    thrice = i * 3
    digits = str(i) + str(twice) + str(thrice)

    if len(digits) != 9:
      continue
    if len(set(digits)) == len(digits):
      pandigitals.append(int(digits))

  for i in range(5000,9999):
    if i % 5 == 0:
      continue
    twice = i * 2
    digits = str(i) + str(twice)

    if len(digits) != 9:
      continue
    if len(set(digits)) == len(digits) and '0' not in digits:
      pandigitals.append(int(digits))

  return max(pandigitals)

pandigital_product()
```

## Problem 39

<p>If $p$ is the perimeter of a right angle triangle with integral length sides, $\{a, b, c\}$, there are exactly three solutions for $p = 120$.</p>
<p>$\{20,48,52\}$, $\{24,45,51\}$, $\{30,40,50\}$</p>
<p>For which value of $p \le 1000$, is the number of solutions maximized?</p>


```python
import math
def pythagorean_triples(limit):
  values = [0] * limit
  for i in range(2,23): # 23 is max because we need i^2 + ij <= 500
    for j in range(1,i):
      if math.gcd(i,j) != 1 or (i - j) % 2 == 0: # conditions for primitive pythagorean triples
        continue
      count = 1
      a = i * i - j * j
      b = 2 * i * j
      c = i * i + j * j
      #print(a,b,c)
      triple_sum = a + b + c
      while triple_sum <= len(values):
        values[triple_sum - 1] += 1
        count += 1
        triple_sum = count * a + count * b + count * c

  return values.index(max(values)) + 1

pythagorean_triples(1000)
```

## Problem 40

<p>An irrational decimal fraction is created by concatenating the positive integers:
$$0.123456789101112131415161718192021\cdots$$</p>
<p>It can be seen that the $12$<sup>th</sup> digit of the fractional part is $1$.</p>
<p>If $d_n$ represents the $n$<sup>th</sup> digit of the fractional part, find the value of the following expression.
$$d_1 \times d_{10} \times d_{100} \times d_{1000} \times d_{10000} \times d_{100000} \times d_{1000000}$$</p>

```python
#Champernowne's Constant
champ = []
for i in range(1,1000000):
  champ.append(str(i))
champ = ''.join(champ)
product = 1
for j in range(7):
  product *= int(champ[10**j - 1])

print(product)
```
