# Problem 71

Consider the fraction, $\dfrac{n}{d}$, where $n$ and $d$ are positive integers. If $n < d$ and $\operatorname{HCF}(n,d)=1$, it is called a reduced proper fraction.

If we list the set of reduced proper fractions for $d \le 8$ in ascending order of size, we get:

$$\frac{1}{8}, \frac{1}{7}, \frac{1}{6}, \frac{1}{5}, \frac{1}{4}, \frac{2}{7}, \frac{1}{3}, \frac{3}{8}, \mathbf{\frac{2}{5}}, \frac{3}{7}, \frac{1}{2}, \frac{4}{7}, \frac{3}{5}, \frac{5}{8}, \frac{2}{3}, \frac{5}{7}, \frac{3}{4}, \frac{4}{5}, \frac{5}{6}, \frac{6}{7}, \frac{7}{8}$$

It can be seen that $\dfrac{2}{5}$ is the fraction immediately to the left of $\dfrac{3}{7}$.

By listing the set of reduced proper fractions for $d \le 1{,}000{,}000$ in ascending order of size, find the numerator of the fraction immediately to the left of $\dfrac{3}{7}$.

```python
import math
fracs = []
for d in range(8, 1000001):
    n = math.floor(d*(3/7))
    if math.gcd(n,d) == 1:
        fracs.append((n/d, str(n), str(d)))


sorted_fracs = sorted(fracs)
print(sorted_fracs[-1][1], sorted_fracs[-1][2])
```

# Problem 72

Consider the fraction, $\dfrac n d$, where $n$ and $d$ are positive integers. If $n \lt d$ and $\operatorname{HCF}(n,d)=1$, it is called a reduced proper fraction.
If we list the set of reduced proper fractions for $d \le 8$ in ascending order of size, we get:
$$\frac 1 8, \frac 1 7, \frac 1 6, \frac 1 5, \frac 1 4, \frac 2 7, \frac 1 3, \frac 3 8, \frac 2 5, \frac 3 7, \frac 1 2, \frac 4 7, \frac 3 5, \frac 5 8, \frac 2 3, \frac 5 7, \frac 3 4, \frac 4 5, \frac 5 6, \frac 6 7, \frac 7 8$$
It can be seen that there are $21$ elements in this set.
How many elements would be contained in the set of reduced proper fractions for $d \le 1\,000\,000$?


```python
import numpy as np

def prime_factors(n):
    factors = set()

    if n % 2 == 0:
        factors.add(2)
        while n % 2 == 0:
            n //= 2

    i = 3
    while i * i <= n:
        if n % i == 0:
            factors.add(i)
            while n % i == 0:
                n //= i
        i += 2

    if n > 1:
        factors.add(n)

    return factors # not including 1 and self

def totient(n):
    factors = prime_factors(n)

    for p in factors:
        n *= (1 - 1/p)

    return n # this is phi(n)
```

```python
frac_count = 0

for d in range(2, 1000001):
    frac_count += totient(d)
        
print(int(frac_count))
```

# Problem 73

Consider the fraction, $\dfrac{n}{d}$, where $n$ and $d$ are positive integers. If $n < d$ and $\operatorname{HCF}(n,d)=1$, it is called a reduced proper fraction.

If we list the set of reduced proper fractions for $d \le 8$ in ascending order of size, we get:

$$\frac{1}{8}, \frac{1}{7}, \frac{1}{6}, \frac{1}{5}, \frac{1}{4}, \frac{2}{7}, \frac{1}{3}, \mathbf{\frac{3}{8}}, \mathbf{\frac{2}{5}}, \mathbf{\frac{3}{7}}, \frac{1}{2}, \frac{4}{7}, \frac{3}{5}, \frac{5}{8}, \frac{2}{3}, \frac{5}{7}, \frac{3}{4}, \frac{4}{5}, \frac{5}{6}, \frac{6}{7}, \frac{7}{8}$$

It can be seen that there are $3$ fractions between $\dfrac{1}{3}$ and $\dfrac{1}{2}$.

How many fractions lie between $\dfrac{1}{3}$ and $\dfrac{1}{2}$ in the sorted set of reduced propert fractions for $d \le 12{,}000$?

```python
import math
fracs = []
for d in range(4, 12001): # 2 and 3 don't round for 1/2 and 1/3
    n_upper = math.floor(d*(1/2))
    n_lower = math.ceil(d*(1/3))
    for n in range(n_lower, n_upper + 1):
        if math.gcd(n,d) == 1:
            fracs.append((n/d, str(n), str(d)))


print(len(fracs))
```

# Problem 74

The number $145$ is well known for the property that the sum of the factorial of its digits is equal to $145$:
$$1! + 4! + 5! = 1 + 24 + 120 = 145.$$
Perhaps less well known is $169$, in that it produces the longest chain of numbers that link back to $169$; it turns out that there are only three such loops that exist:
$$\begin{align*}
&169 \to 363601 \to 1454 \to 169\\
&871 \to 45361 \to 871\\
&872 \to 45362 \to 872
\end{align*}$$
It is not difficult to prove that EVERY starting number will eventually get stuck in a loop. For example,
$$\begin{align*}
&69 \to 363600 \to 1454 \to 169 \to 363601 (\to 1454)\\
&78 \to 45360 \to 871 \to 45361 (\to 871)\\
&540 \to 145 (\to 145)
\end{align*}$$
Starting with $69$ produces a chain of five non-repeating terms, but the longest non-repeating chain with a starting number below one million is sixty terms.
How many chains, with a starting number below one million, contain exactly sixty non-repeating terms?


```python
import numpy as np
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n-1)
    
def fact_digit_sum(m):
    return sum(factorial(int(digit)) for digit in str(m))

number = 0
steps_away = {145: 0, 169: 3, 871: 2, 872: 2, 40585: 0} # these are the known loops to reference

for i in range(1, 1000001):
    if i in steps_away:
        continue
    chain = []
    fact_sum = i
    count = 0

    while fact_sum not in steps_away and fact_sum not in chain:
        chain.append(fact_sum)
        fact_sum = fact_digit_sum(fact_sum)
        
    #print(chain)
    steps_away[i] = len(chain) if chain else steps_away[fact_sum]
    steps = steps_away[fact_sum]

    for j, num in enumerate(reversed(chain)):
        steps_away[num] = steps + j + 1

    if steps_away[i] == 60:
        number += 1

print(number)
```
