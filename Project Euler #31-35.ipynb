<h1>Problem 31</h1>
The British pound has eight coins in general circulation: 1p, 2p, 5p, 10p, 20p, 50p, 100p (1 pound), and 200p (2 pounds). How many different ways are there to make 2 pounds with any number of coins?

```python
def findchange(coins, target):
    
    coinlist = [0] * (target + 1)
    coinlist[0] = 1        # 1 way to make 0 coins
    
    for coin in coins:
        for i in range(coin, target + 1):
            coinlist[i] += coinlist[i-coin]   # use this to show how many ways there are to hit target after using the current coin first
            
    return coinlist[target]

findchange([1,2,5,10,20,50,100,200], 200)
```

<h1>Problem 32</h1>
An $n$-digit number is pandigital if it contains all the numbers 1 to $n$ exactly once. The product 7254 is unusual because the identity, $39\times 186 = 7254$, containing the multiplicand, multiplier, and product is 1 through 9 pandigital. Find the sum of all products whose multiplicand/multiplier/product identity can be written as a 1 through 9 pandigital.

```python
pandigitals = []

for i in range(2,100):
    for j in range(101,2000):
        product = i * j
        if product > 9999:
            break
        numbers = [i, j, product]
        number_string = ''.join(map(str, numbers))
        digits = list(number_string)
        no_zeros = [digit for digit in digits if digit != '0']
        if len(set(no_zeros)) == 9:
            pandigitals.append(product)
            
pandigitals = set(pandigitals)
print(sum(pandigitals))
```

<h1>Problem 33</h1>
An inexperienced mathematician might incorrectly believe that the fraction $\frac{49}{98}=\frac48$ is obtained by cancelling the $9$s. We consider cases like $\frac{30}{50}=\frac35$ to be trivial examples. There are exactly four non-trivial examples of this type of fraction, less than one in value, and containing two digits in the numerator and the denominator. If the product of these four fractions is given in its lowest common terms, find the value of the denominator.

```python
from fractions import Fraction

fractionlist = []
for den in range(12,100):
    for num in range(11,den):

        frac = num / den
        
        num_1st_digit = num // 10
        num_2nd_digit = num % 10
        den_1st_digit = den // 10
        den_2nd_digit = den % 10
        
        
        if num_2nd_digit != 0 and den_2nd_digit != 0:
        
            if num_1st_digit == den_1st_digit:
                if num_2nd_digit / den_2nd_digit == frac:
                    fractionlist.append(num)
                    fractionlist.append(den)
            elif num_1st_digit == den_2nd_digit:
                if num_2nd_digit / den_1st_digit == frac:
                    fractionlist.append(num)
                    fractionlist.append(den)
            elif num_2nd_digit == den_1st_digit:
                if num_1st_digit / den_2nd_digit == frac:
                    fractionlist.append(num)
                    fractionlist.append(den)
            elif num_2nd_digit == den_2nd_digit:
                if num_1st_digit / den_1st_digit == frac:
                    fractionlist.append(num)
                    fractionlist.append(den)
                
product = 1
for i in range(0,len(fractionlist),2):
    product *= (fractionlist[i] / fractionlist[i+1])
    
print(Fraction(product).limit_denominator(1000))
```

<h1>Problem 34</h1>
Find the sum of all numbers which are equal to the sum of the factorial of their digits.

```python
import math

tot_sum = 0

for i in range(10, math.factorial(9) * 7): # upper bound is because a seven digit number is the maximum that could equal its digit's factorial sum
    fact_sum = 0
    n_string = str(i)
    for char in n_string:
        fact_sum += math.factorial(int(char))
    if fact_sum == i:
        tot_sum += fact_sum
        #print(i)
print(tot_sum)
```
