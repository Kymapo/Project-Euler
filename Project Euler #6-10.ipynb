<h1>Problem 6</h1>
Find the difference between the sum of the squares of the first one hundred natural numbers and the square of the sum.

```python
def sum_of_squares():
    sum_square = 0
    for i in range(1,101):
        sum_square += i * i
    return sum_square

def squared_sum():
    square_sum = 0
    for j in range(1,101):
        square_sum += j
    return square_sum * square_sum

squared_sum() - sum_of_squares()
```

<h1>Problem 7</h1>
What is the 10001st prime number?

```python
primes = []
def is_prime(number, divisor=None):
    if divisor is None:
        divisor = 2
    if number <= 1:
        return False
    if divisor * divisor > number:
        return True
    if number % divisor == 0:
        return False
    return is_prime(number, divisor+1)

y = 0
while len(primes) < 10001:
    if is_prime(y) == True:
        primes.append(y)
    y += 1

print(primes[-1])
```

<h1>Problem 9</h1>
There exists exactly one Pythagorean triplet for which $a+b+c=1000$. Find the product $abc$.

```python
#  We will use the shortcut for finding Pythagorean triplets, a=m^2-n^2, b=2mn, and c=m^2+n^2 for integers m>n.

def pythagorean_triplet():
    for n in range(1,15): #found 15 with some algebra
        m=n+1
        while 2*(m*m) + 2*m*n <= 1000:
            if 2*(m*m) + 2*m*n == 1000:
                print('a=',m*m - n*n,'b=',2*m*n,'c=',m*m + n*n)
                return (m*m - n*n) * (2*m*n) * (m*m + n*n)
            m = m + 1
            
pythagorean_triplet()
```

<h1>Problem 10</h1>
Find the sum of all primes below 2 million.

```python
primes = [2]
def is_prime(number, divisor=None):
    if divisor is None:
        divisor = 2
    if number <= 1:
        return False
    if divisor * divisor > number:
        return True
    if number % divisor == 0:
        return False
    return is_prime(number, divisor+1)

y = 0
i = 3
while primes[y] < 2000000:
    if is_prime(i) == True:
        primes.append(i)
        y = y + 1
    i = i + 1

primes.pop()
print(sum(primes))
```
