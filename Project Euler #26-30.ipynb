<h1>Problem 26</h1>
Find the value of $d<1000$ for which $1/d$ has the longest recurring cycle in its decimal fraction part.

```python
def cycle_length(denom):
    numer = 1
    remainders = []
    while len(remainders) == len(set(remainders)):
        remainder = numer % denom
        remainders.append(remainder)
        numer = remainder * 10
    return len(set(remainders))

curr_max = 0
for i in range(1,1001,2):
    if cycle_length(i) > curr_max:
        curr_max = i
        
print(curr_max)
```

Euler's formula $n^2 + n + 41$ produces 40 primes for the consecutive integers $0 \leq n \leq 39$. Considering quadratics of the form $n^2 + an + b$ where $|a| < 1000$ and $|b| \leq 1000$, find the integers a and b that produces the maximum number of consecutive integers starting at $n=0$.

```python
primes = [2]
def is_prime(number, divisor=None):
    if divisor is None:
        divisor = 2
    if number <= 1:
        return False
    if divisor * divisor > number:
        return True
    if number % divisor == 0:
        return False
    return is_prime(number, divisor+1)

y = 3
while primes[-1] < 1001:
    if is_prime(y) == True:
        primes.append(y)
    y += 1

primes.pop()

negative_primes = [-1 * p for p in primes]

max_len = 0
max_a = 0
max_b = 0

def calc_seq(a,b):
    i = -1
    output = 2
    while is_prime(output) == True:
        i = i + 1
        output = i * i + a * i + b
    return i

for i in primes:
    for j in primes:                
        if calc_seq(i,j) > max_len:
            max_a = i
            max_b = j
            max_len = calc_seq(i,j)
        if calc_seq(i,-j) > max_len:
            max_a = i
            max_b = -j
            max_len = calc_seq(i,-j)
        if calc_seq(-i,j) > max_len:
            max_a = -i
            max_b = j
            max_len = calc_seq(-i,j)
        if calc_seq(-i,-j) > max_len:
            max_a = -i
            max_b = -j
            max_len = calc_seq(-i,-j)
        
print('a=', max_a, 'b=', max_b, 'product=', max_a * max_b, 'length=', max_len)
```

<h1>Problem 28</h1>
What is the sum of the numbers on the diagonals of a 1001 by 1001 spiral formed moving to the right and clockwise starting with 1 in ascending order?

```python
import numpy as np

spiral = np.zeros((1001,1001))     # change size here

counter = 0
x = spiral.shape[0] // 2
y = spiral.shape[0] // 2
spiral[x][y] = counter + 1
spiral[x][y+1] = counter + 2
spiral[x+1][y+1] = counter + 3
counter += 4

for i in range(1, x):
        spiral[x+i][y-i:y+i] = range(counter + 2*i-1, counter - 1, -1) 
        counter += 2*i
        spiral[x-i:x+i,y-i] = np.array(range(counter + 2*i-1, counter - 1, -1))
        counter += 2*i
        
        spiral[x-i][y+1-i:y+1+i+1] = range(counter, counter + 2*i+1)
        counter += 2*i + 1
        spiral[x-i+1:x+i+2,y+i+1] = np.array(range(counter, counter + 2*i + 1))
        counter += 2*i + 1

i+=1
spiral[x+i][y-i:y+i] = range(counter + 2*i-1, counter - 1, -1) 
counter += 2*i
spiral[x-i:x+i,y-i] = np.array(range(counter + 2*i-1, counter - 1, -1))
counter += 2*i
spiral[0][1:spiral.shape[0]] = range(counter, counter + spiral.shape[0]-1)

nsum = 0
for j in range(spiral.shape[0]):
    nsum += spiral[j][j]
    nsum += spiral[j][spiral.shape[0] - j - 1]
    
nsum -= 1
print(nsum)
```

<h1>Problem 29</h1>
How many distinct terms are in the sequence generated by $a^b$ for $2\leq a \leq 100$ and $2\leq b \leq 100$?

```python
import numpy as np

array = np.zeros((99,99))

for i in range(array.shape[0]):
    for j in range(array.shape[0]):
        array[i][j] = (i+2) ** (j+2)
        
sortednlist = sorted(np.unique(array))
print(len(sortednlist))
```

<h1>Problem 30</h1>
Find the sum of all numbers that can be written as the sum of the fifth powers of its digits.

```python
nlist = []
for i in range (1000,300000):
    numsum = 0
    digits = str(i)
    for num in digits:
        numsum += int(num) ** 5
    
    if numsum == i:
        nlist.append(i)
        
print(nlist)
print(sum(nlist))
```
