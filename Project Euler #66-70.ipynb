# Problem 66

Consider quadratic Diophantine equations of the form:
$$x^2 - Dy^2 = 1$$

For example, when $D=13$, the minimal solution in $x$ is $649^2 - 13 \times 180^2 = 1$.

It can be assumed that there are no solutions in positive integers when $D$ is square.

By finding minimal solutions in $x$ for $D = \{2, 3, 5, 6, 7\}$, we obtain the following:

$$\begin{align}
3^2 - 2 \times 2^2 &= 1\\
2^2 - 3 \times 1^2 &= 1\\
\color{red}{\mathbf{9}}^2 - 5 \times 4^2 &= 1\\
5^2 - 6 \times 2^2 &= 1\\
8^2 - 7 \times 3^2 &= 1
\end{align}$$

Hence, by considering minimal solutions in $x$ for $D \le 7$, the largest $x$ is obtained when $D=5$.

Find the value of $D \le 1000$ in minimal solutions of $x$ for which the largest value of $x$ is obtained.

````python
import math
from decimal import Decimal, getcontext
def calc_period(n, seq, frac_seq=[]):
    
    count = 0
    index = 0
    while True:
        
        if count == 0:
            fraction = Decimal(1) / (Decimal(math.sqrt(n)) - math.floor(Decimal(math.sqrt(n))))
            frac_seq.append(fraction)
            frac_num = seq[-1]

            frac_int = math.floor(frac_seq[-1])
            frac_den = n - frac_num**2
            frac_num = frac_int * frac_den - frac_num
        
        else:
            frac_int = math.floor(frac_seq[-1])
            if frac_den != 1:
                frac_den = (n - frac_num**2) // frac_den
            else:
                frac_den = n - frac_num**2
            frac_num = frac_int * frac_den - frac_num

        fraction = Decimal(1 / ((math.sqrt(n) - frac_num) / frac_den))

        if fraction in frac_seq:
            index = frac_seq.index(fraction)
            period = len(seq) - index
            break
        
        frac_seq.append(fraction)

        seq.append(frac_int)

        count += 1

    return seq, frac_seq, period
````

````python
def find_min_sol(D):

    if math.sqrt(D) % 1 != 0:
        seq, _, period = calc_period(D, [math.isqrt(D)], frac_seq=[])

    a_k = seq
    get_a_k = lambda k: a_k[0] if k == 0 else a_k[1 + (k - 1) % period] # loop with period

    p, q, j = [1, get_a_k(0)], [0, 1], 1 # starting recurrents
    while True:
        p.append(get_a_k(j) * p[-1] + p[-2])
        q.append(get_a_k(j) * q[-1] + q[-2])

        if p[-1] * p[-1] - D * q[-1] * q[-1] == 1:
            return (p[-1], D, q[-1]) # (x, D, y)

        j += 1

Max = 0
max_D = 0
for D in range(2, 1001):
    if math.isqrt(D)**2 != D:
        x, _, _ = find_min_sol(D)
        if x > Max:
            Max = x
            max_D = D

print(max_D)
````

# Problem 67

By starting at the top of the triangle below and moving to adjacent numbers on the row below, the maximum total from top to bottom is 23.
```
    **3**
   **7** 4
  2 **4** 6
 8 5 **9** 3
```

That is, 3 + 7 + 4 + 9 = 23.

Find the maximum total from top to bottom in [triangle.txt](resources/documents/0067_triangle.txt) (right click and 'Save Link/Target As...'), a 15K text file containing a triangle with one-hundred rows.

**NOTE:** This is a much more difficult version of [Problem 18](problem=18). It is not possible to try every route to solve this problem, as there are $2^{99}$ altogether! If you could check one trillion ($10^{12}$) routes every second it would take over twenty billion years to check them all. There is an efficient algorithm to solve it. ;o)

````python
with open('0067_triangle.txt', 'r') as f:
    triangle = [row.split() for row in f.readlines()]
    triangle = [[int(x) for x in row] for row in triangle]

def find_max_path(triangle):

    for i in range(len(triangle) - 1, 0, -1):
        for j in range(len(triangle[i]) - 1):
            triangle[i-1][j] += max(triangle[i][j], triangle[i][j+1])

    return triangle[0][0]

print(find_max_path(triangle))
````

<p>Using the numbers 1 to 10, and depending on arrangements, it is possible to form 16- and 17-digit strings. What is the maximum <b>16-digit</b> string for a "magic" 5-gon ring?</p>

````python
import numpy as np
from itertools import permutations

nums = [i for i in range(1, 11)]
sols = []

def make_5gon(gons, row, S, nums):

    if row == gons.shape[0]:
        if gons[-1, 2] == gons[0, 1] and len(set(gons.flatten())) == 10:
            sols.append((S, gons.copy()))
            return True
        return False
    
    chosen = set(gons[:row, :].flatten())
    left = set(nums) - chosen
    left.add(gons[0, 1])

    gons[row, 1] = gons[row-1, 2]
    #print(gons)

    for perm in permutations(left, 2):
        #print("Trying permutation:", perm)
        #print(left)
        #print(gons)
        if sum(perm) == S - gons[row, 1]:
            gons[row, 0], gons[row, 2] = perm
        

            make_5gon(gons, row + 1, S, nums)
                #print(gons)
                #return True
            
            gons[row, 0], gons[row, 2] = 0, 0

    return False



for S in range(10, 21):
    for perm in (p for p in permutations(nums[:-1], 2) if sum(p) == S - 10):
        #print("Trying with starting pair:", perm)
        gons = np.zeros((5, 3), dtype=int)
        gons[0, 0] = 10
        gons[0, 1:] = perm
        #print(gons)

        make_5gon(gons, 1, S, nums)


def find_max_string(sol):
    _, gons = sol
    min_index = np.argmin(gons[:, 0]) # find the index of the minimum external node for max string
    rotated = np.roll(gons, -min_index, axis=0)
    return ''.join(map(str, rotated.flatten()))

max_string = ''
for sol in sols:
    max_string = max(max_string, find_max_string(sol))
print(max_string)
````
