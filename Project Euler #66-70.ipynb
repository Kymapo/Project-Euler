# Problem 66

Consider quadratic Diophantine equations of the form:
$$x^2 - Dy^2 = 1$$

For example, when $D=13$, the minimal solution in $x$ is $649^2 - 13 \times 180^2 = 1$.

It can be assumed that there are no solutions in positive integers when $D$ is square.

By finding minimal solutions in $x$ for $D = \{2, 3, 5, 6, 7\}$, we obtain the following:

$$\begin{align}
3^2 - 2 \times 2^2 &= 1\\
2^2 - 3 \times 1^2 &= 1\\
\color{red}{\mathbf{9}}^2 - 5 \times 4^2 &= 1\\
5^2 - 6 \times 2^2 &= 1\\
8^2 - 7 \times 3^2 &= 1
\end{align}$$

Hence, by considering minimal solutions in $x$ for $D \le 7$, the largest $x$ is obtained when $D=5$.

Find the value of $D \le 1000$ in minimal solutions of $x$ for which the largest value of $x$ is obtained.

````python
import math
from decimal import Decimal, getcontext
def calc_period(n, seq, frac_seq=[]):
    
    count = 0
    index = 0
    while True:
        
        if count == 0:
            fraction = Decimal(1) / (Decimal(math.sqrt(n)) - math.floor(Decimal(math.sqrt(n))))
            frac_seq.append(fraction)
            frac_num = seq[-1]

            frac_int = math.floor(frac_seq[-1])
            frac_den = n - frac_num**2
            frac_num = frac_int * frac_den - frac_num
        
        else:
            frac_int = math.floor(frac_seq[-1])
            if frac_den != 1:
                frac_den = (n - frac_num**2) // frac_den
            else:
                frac_den = n - frac_num**2
            frac_num = frac_int * frac_den - frac_num

        fraction = Decimal(1 / ((math.sqrt(n) - frac_num) / frac_den))

        if fraction in frac_seq:
            index = frac_seq.index(fraction)
            period = len(seq) - index
            break
        
        frac_seq.append(fraction)

        seq.append(frac_int)

        count += 1

    return seq, frac_seq, period
````

````python
def find_min_sol(D):

    if math.sqrt(D) % 1 != 0:
        seq, _, period = calc_period(D, [math.isqrt(D)], frac_seq=[])

    a_k = seq
    get_a_k = lambda k: a_k[0] if k == 0 else a_k[1 + (k - 1) % period] # loop with period

    p, q, j = [1, get_a_k(0)], [0, 1], 1 # starting recurrents
    while True:
        p.append(get_a_k(j) * p[-1] + p[-2])
        q.append(get_a_k(j) * q[-1] + q[-2])

        if p[-1] * p[-1] - D * q[-1] * q[-1] == 1:
            return (p[-1], D, q[-1]) # (x, D, y)

        j += 1

Max = 0
max_D = 0
for D in range(2, 1001):
    if math.isqrt(D)**2 != D:
        x, _, _ = find_min_sol(D)
        if x > Max:
            Max = x
            max_D = D

print(max_D)
````
