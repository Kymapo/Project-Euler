Find the minimal path sum from the left column to the right column in <a href="resources/documents/0082_matrix.txt">matrix.txt</a> (right click and "Save Link/Target As..."), a 31K text file containing an $80$ by $80$ matrix.

```python
import copy

with open("0081_matrix.txt", "r") as f:
    matrix = [f.readline().strip().split(",") for _ in range(80)]
matrix_down = copy.deepcopy(matrix)
matrix_up = copy.deepcopy(matrix)


for m in range(78, -1, -1):
    
    matrix_down[79][m] = int(matrix_down[79][m]) + int(matrix_down[79][m+1])
    for j in range(78, -1, -1): # looking below
        matrix_down[j][m] = min(int(matrix[j][m]) + int(matrix_down[j+1][m]), int(matrix_down[j][m]) + int(matrix_down[j][m+1]))

    matrix_up[0][m] = int(matrix_up[0][m]) + int(matrix_up[0][m+1])
    for k in range(1, 80): # looking above
        matrix_up[k][m] = min(int(matrix[k][m]) + int(matrix_up[k-1][m]), int(matrix_up[k][m]) + int(matrix_up[k][m+1]))

    for i in range(80):
        matrix_down[i][m] = min(matrix_down[i][m], matrix_up[i][m])
        matrix_up[i][m] = min(matrix_down[i][m], matrix_up[i][m])

print(min(row[0] for row in matrix_down)) # same as matrix_up
```

# Problem 83

NOTE: This problem is a significantly more challenging version of <a href="problem=81">Problem 81</a>.
In the $5$ by $5$ matrix below, the minimal path sum from the top left to the bottom right, by moving left, right, up, and down, is indicated in bold red and is equal to $2297$.
$$
\begin{pmatrix}
\color{red}{131} & 673 & \color{red}{234} & \color{red}{103} & \color{red}{18}\\
\color{red}{201} & \color{red}{96} & \color{red}{342} & 965 & \color{red}{150}\\
630 & 803 & 746 & \color{red}{422} & \color{red}{111}\\
537 & 699 & 497 & \color{red}{121} & 956\\
805 & 732 & 524 & \color{red}{37} & \color{red}{331}
\end{pmatrix}
$$
Find the minimal path sum from the top left to the bottom right by moving left, right, up, and down in <a href="resources/documents/0083_matrix.txt">matrix.txt</a> (right click and "Save Link/Target As..."), a 31K text file containing an $80$ by $80$ matrix.

```python
import heapq

def dijkstra_in_matrix(matrix, start, end):
    rows, cols = len(matrix), len(matrix[0])
    directions = [(0, 1), (1, 0), (0, -1), (-1, 0)]  # right, down, left, up

    min_heap = [(int(matrix[start[0]][start[1]]), start)] # cost, (row, column)
    distances = {}

    while min_heap:
        cost, (r, c) = heapq.heappop(min_heap) # row, column

        if (r, c) in distances:
            continue # means a shorter path was already found
        distances[(r, c)] = cost
        if (r, c) == end:
            return cost
        
        for dr, dc in directions:
            nr, nc = r + dr, c + dc
            if 0 <= nr < rows and 0 <= nc < cols and (nr, nc) not in distances:
                heapq.heappush(min_heap, (cost + int(matrix[nr][nc]), (nr, nc)))

    return float('inf')  # if end is not reachable

with open("0081_matrix.txt", "r") as f:
    matrix = [f.readline().strip().split(",") for _ in range(80)]

dijkstra_in_matrix(matrix, (0, 0), (79, 79))
```

# Problem 84

In the game, **Monopoly**, the standard board is set up in the following way:

![Monopoly Board](resources/images/0084_monopoly_board.png)

A player starts on the GO square and adds the scores on two 6-sided dice to
determine the number of squares they advance in a clockwise direction. Without
any further rules we would expect to visit each square with equal probability:
2.5%. However, landing on G2J (Go To Jail), CC (community chest), and CH
(chance) changes this distribution.

In addition to G2J, and one card from each of CC and CH, that orders the player
to go directly to jail, if a player rolls three consecutive doubles, they do not
advance the result of their 3rd roll. Instead they proceed directly to jail.

At the beginning of the game, the CC and CH cards are shuffled. When a player
lands on CC or CH they take a card from the top of the respective pile and,
after following the instructions, it is returned to the bottom of the pile.
There are sixteen cards in each pile, but for the purpose of this problem we
are only concerned with cards that order a movement; any instruction not
concerned with movement will be ignored and the player will remain on the CC/CH
square.

- **Community Chest (2/16 cards):**
  1. Advance to GO
  2. Go to JAIL

- **Chance (10/16 cards):**
  1. Advance to GO
  2. Go to JAIL
  3. Go to C1
  4. Go to E3
  5. Go to H2
  6. Go to R1
  7. Go to next R (railway company)
  8. Go to next R
  9. Go to next U (utility company)
  10. Go back 3 squares.

The heart of this problem concerns the likelihood of visiting a particular
square. That is, the probability of finishing at that square after a roll. For
this reason it should be clear that, with the exception of G2J for which the
probability of finishing on it is zero, the CH squares will have the lowest
probabilities, as 5/8 request a movement to another square, and it is the final
square that the player finishes at on each roll that we are interested in. We
shall make no distinction between "Just Visiting" and being sent to JAIL, and
we shall also ignore the rule about requiring a double to "get out of jail",
assuming that they pay to get out on their next turn.

By starting at GO and numbering the squares sequentially from 00 to 39 we can
concatenate these two-digit numbers to produce strings that correspond with sets
of squares.

Statistically it can be shown that the three most popular squares, in order, are
JAIL (6.24%) = Square 10, E3 (3.18%) = Square 24, and GO (3.09%) = Square 00.
So these three most popular squares can be listed with the six-digit modal
string: **102400**.

If, instead of using two 6-sided dice, two 4-sided dice are used, find the
six-digit modal string.

```python
import random
import heapq
space_counts = [0] * 40
space_probs = [0] * 40
space = 0 # start at GO
space_counts[space] += 1
go = 0 # space numbers
jail = 10
g2j = 30
rr = [5, 15, 25, 35] # railroad
u = [12, 28] # utility
cc = [2, 17, 33]
cc_cards = [-1] * 16 # set -1 for nothing to avoid space number 0 issues
cc_cards[0] = go
cc_cards[1] = jail
random.shuffle(cc_cards)
ch = [7, 22, 36]
ch_cards = [-1] * 16
ch_cards[0] = go
ch_cards[1] = jail
ch_cards[2] = 11 # C1
ch_cards[3] = 24 # E3
ch_cards[4] = 39 # H2
ch_cards[5] = 5 # R1
ch_cards[6] = "next_rr"
ch_cards[7] = "next_rr"
ch_cards[8] = "next_u"
ch_cards[9] = "back_3"
random.shuffle(ch_cards)
consecutive_doubles = 0

for _ in range(3000000):
    die1 = random.randint(1, 4)
    die2 = random.randint(1, 4)
    roll = die1 + die2
    if die1 == die2:
        consecutive_doubles += 1
    else:
        consecutive_doubles = 0 # reset doubles
    if consecutive_doubles == 3: # go to jail
        space = jail
        consecutive_doubles = 0 # reset
    else:
        space = (space + roll) % 40 # going around the board if not in jail


    if space == g2j: # go to jail
        space = jail
    elif space in cc: # community chest
        card = cc_cards[0] # draw the top card
        cc_cards = cc_cards[1:] + [card] # put it at the bottom
        if card != -1: # cc doesn't need string check
            space = card
    elif space in ch: # chance
        card = ch_cards[0]
        ch_cards = ch_cards[1:] + [card]
        if card != -1:
            if isinstance(card, str): # variable space cards
                if card == "next_rr":
                    space = next((r for r in rr if r > space), rr[0]) # go to next railroad
                elif card == "next_u":
                    space = next((u for u in u if u > space), u[0]) # go to next utility
                elif card == "back_3":
                    space = (space - 3) % 40 # go back 3 spaces
            else:
                space = card # go to the space on the card if it's a number
                
    space_counts[space] += 1 # count space AFTER any event
    space_probs[space] = space_counts[space] / sum(space_counts)

# Get the 3 most common spaces
top3 = heapq.nlargest(3, enumerate(space_probs), key=lambda x: x[1])
for i, (space, prob) in enumerate(top3):
    print(f"Space {space}: {prob:.4f}")
```
