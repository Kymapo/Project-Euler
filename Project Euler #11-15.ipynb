<h1>Problem 14</h1>
Which starting number, under one million, produces the longest Collatz Sequence?
$ n \rightarrow \frac{n}{2} \ \  \text{if n is even,}$
$n \rightarrow 3n+1 \ \ \text{if n is odd}$

```python
def collatz_seq(seq):
    i = 0
    while seq[i] != 1:
        if seq[i] % 2 == 0:
            seq.append(int(seq[i] / 2))
        else:
            seq.append(3 * seq[i] + 1)
        i += 1
    return seq
        
max_seq = []
for i in range(500000, 1000000):
    collatz = [i]
    number_seq = collatz_seq(collatz)
    if len(number_seq) > len(max_seq):
        max_seq = number_seq
        
#print(len(max_seq))
print(max_seq[0])
```

<h1>Problem 15</h1>
Starting at the top left corner of a 20x20 grid and only being able to move right and down, how many different routes are there to the bottom right corner?

```python
import numpy as np

def find_paths(num):
    num = num + 1
    grid = []
    for i in range(0, num):
        grid.append([1] * num) # need this way to not cause error
    grid[0][0] = 0
    for j in range(1, num):
        for k in range(1, num):
            grid[j][k] = grid[j-1][k] + grid[j][k-1]
    return grid[-1][-1]
find_paths(20)
```
