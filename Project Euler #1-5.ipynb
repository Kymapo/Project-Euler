<h1>Problem 1</h1>
Find the sum of all multiples of 3 or 5 below 1000.

```python
multiples = []
for i in range(1000):
    if i % 3 == 0 or i % 5 == 0:
        multiples.append(i)

multiples_sum = sum(multiples)
print(multiples_sum)
```

What is the largest prime factor of the number 600851475143?

<h1>Problem 4</h1>
Find the largest palindromic number made from the product of two 3-digit numbers.

```python
from sympy import divisors

palindromes = []

for i in range(1,10):
    for j in range(0,10):
        for k in range(0,10):
            palindromes.append((i*100000 + i) + (j*10000 + j*10) + (k*1000 + k*100))

def find_palindrome():
    for z in reversed(palindromes):
        #print(divisors(z))
        for j in range (0, int(len(divisors(z)) / 2)):
            #print(divisors(z)[j], divisors(z)[-1*(j+1)])
            if divisors(z)[j] >= 100 and divisors(z)[-1*(j+1)] < 1000:
                return z

find_palindrome()
```

<h1>Problem 5</h1>
What is the smallest positive number that is evenly divisible by all of the numbers 1 through 20?

```python
import math
from sympy import divisors

def find_num():
    for x in divisors(math.factorial(20)):
        if divisors(x)[0:20] == list(range(1,21)):
            return x
        
find_num()
```
