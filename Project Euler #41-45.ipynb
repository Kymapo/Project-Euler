## Problem 41

<p>We shall say that an $n$-digit number is pandigital if it makes use of all the digits $1$ to $n$ exactly once. For example, $2143$ is a $4$-digit pandigital and is also prime.</p>
<p>What is the largest $n$-digit pandigital prime that exists?</p>


```python
from sympy import isprime
from itertools import permutations

digits = ['9','8','7','6','5','4','3','2','1']

def pandigital_prime(digits):
  for i in range(len(digits)):
          perms = permutations(digits[i:])
          for pandigital in perms:
              num = int("".join(pandigital))
              if isprime(num):
                  return num

pandigital_prime(digits)
```

## Problem 42

<p>The $n$<sup>th</sup> term of the sequence of triangle numbers is given by, $t_n = \frac12n(n+1)$; so the first ten triangle numbers are:
$$1, 3, 6, 10, 15, 21, 28, 36, 45, 55, \dots$$</p>
<p>By converting each letter in a word to a number corresponding to its alphabetical position and adding these values we form a word value. For example, the word value for SKY is $19 + 11 + 25 = 55 = t_{10}$. If the word value is a triangle number then we shall call the word a triangle word.</p>
<p>Using <a href="resources/documents/0042_words.txt">words.txt</a> (right click and 'Save Link/Target As...'), a 16K text file containing nearly two-thousand common English words, how many are triangle words?</p>

```python
with open('0042_words.txt', 'r') as f:
  words = f.read().split(',')

#print(words)

triangle_nums = []
for i in range(1,25): # sufficient upper bound
  triangle_nums.append(int(0.5 * i * (i + 1)))
triangle_nums = set(triangle_nums)
#print(triangle_nums)

count = 0
for word in words:
  #print(word)
  word_sum = 0
  word_sum = sum((ord(char) - 64) for char in word if char.isalnum())
  #print('sum:', word_sum)
  if word_sum in triangle_nums:
    count += 1

print(count)
```

```python
primes = [2,3,5,7,11,13,17]

def get_multiples(nums):
  multiples_dict = {}
  for num in nums:
    multiples_dict[num] = [f"{i:03}" for i in range(num, 1000, num) if len(set(f"{i:03}")) == len(f"{i:03}")]
  return multiples_dict

numbers = [2,3,5,7,11,13,17]
multiples = get_multiples(numbers)

multiples[13] = [string1 for string1 in multiples[13] if any(string1[-2:] in string2[:2] for string2 in multiples[17])]
multiples[11] = [string1 for string1 in multiples[11] if any(string1[-2:] in string2[:2] for string2 in multiples[13])]
multiples[7] = [string1 for string1 in multiples[7] if any(string1[-2:] in string2[:2] for string2 in multiples[11])]
multiples[5] = [string1 for string1 in multiples[5] if any(string1[-2:] in string2[:2] for string2 in multiples[7])]
multiples[3] = [string1 for string1 in multiples[3] if any(string1[-2:] in string2[:2] for string2 in multiples[5])]
multiples[2] = [string1 for string1 in multiples[2] if any(string1[-2:] in string2[:2] for string2 in multiples[3])]

'''
print(multiples[2])
print(multiples[3])
print(multiples[5])
print(multiples[7])
print(multiples[11])
print(multiples[13])
print(multiples[17])
'''


def find_chains(list1, list2, list3, list4, list5, list6, list7):
  all_chains = []

  def backtrack(current_chain, current_list_index):
        if current_list_index == 7:
            all_chains.append(current_chain.copy())
            return

        current_word = current_chain[-1]
        next_list = [list1, list2, list3, list4, list5, list6, list7][current_list_index]

        for next_word in next_list:
            if current_word[-2:] == next_word[:2]:
                current_chain.append(next_word)
                backtrack(current_chain, current_list_index + 1)  # Recursively explore further with the extended chain
                current_chain.pop()  # Backtrack: Remove the last word to try other paths

  for word1 in list1:
      backtrack([word1], 1)  # Call backtrack with initial chain containing word1 and starting at list2 (index 1)

  return all_chains

all_chains = find_chains(multiples[2], multiples[3], multiples[5], multiples[7], multiples[11], multiples[13], multiples[17])
#print(f"All chains: {all_chains}")
#print(len(all_chains))

full_chains = []
count = 0
for chain in all_chains:
  num = "".join(chain)
  if len(set(num)) == 9:
    full_chains.append(num)
    count += 1

i = 0
for num in full_chains:
  num = num[:3] + num[5] + num[8] + num[11] + num[14] + num[17] + num[20]
  full_chains[i] = num
  i += 1
#print(full_chains)

def find_missing_digit(strings):
  all_digits = '0123456789'
  missing_digits = []

  for string in strings:
    present_digits = set(string)
    missing_digit = next(digit for digit in all_digits if digit not in present_digits)
    missing_digits.append(missing_digit)

  return missing_digits

missing_list = find_missing_digit(full_chains)
#print(missing_list)

final_nums = []

for j in range(len(full_chains)):
  final_nums.append(int(missing_list[j] + full_chains[j]))
print(final_nums)

print(sum(final_nums))
```

## Problem 44

<p>Pentagonal numbers are generated by the formula, $P_n=n(3n-1)/2$. The first ten pentagonal numbers are:
$$1, 5, 12, 22, 35, 51, 70, 92, 117, 145, \dots$$</p>
<p>It can be seen that $P_4 + P_7 = 22 + 70 = 92 = P_8$. However, their difference, $70 - 22 = 48$, is not pentagonal.</p>
<p>Find the pair of pentagonal numbers, $P_j$ and $P_k$, for which their sum and difference are pentagonal and $D = |P_k - P_j|$ is minimised; what is the value of $D$?</p>

```python
pentagonal = lambda x: x * (3*x - 1) / 2
pentagonal_nums = []
min_diff = 1000

for i in range(1, 3000): # not the best way to do this
  pentagonal_nums.append(pentagonal(i))
pentagonal_set = set(pentagonal_nums)

for j in range(1, len(pentagonal_nums)):
  for k in range(j):
    pen_sum = pentagonal_nums[j] + pentagonal_nums[k]
    pen_diff = abs(pentagonal_nums[j] - pentagonal_nums[k])
    if pen_sum in pentagonal_set and pen_diff in pentagonal_set:
      print(f"Pair found: P{k+1} = {pentagonal_nums[k]}, P{j+1} = {pentagonal_nums[j]}")
      print(f"Sum: {pen_sum}, Difference: {pen_diff}")
```
